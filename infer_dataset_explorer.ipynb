```python
from datasets import load_from_disk
import pandas as pd

# 1. Load the dataset (replace with your actual folder name!)
dataset_path = './inference_results_dataset'
print(f"Loading dataset from {dataset_path}...\n")
dataset = load_from_disk(dataset_path)

# 2. Convert to Pandas for easy viewing
df = dataset.to_pandas()

# 3. View basic stats
print(f"Total rows processed: {len(df)}")
print(f"Average time per prompt: {df['time_taken_seconds'].mean():.2f} seconds\n")
```
